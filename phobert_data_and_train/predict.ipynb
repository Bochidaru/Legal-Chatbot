```python
from transformers import AutoModelForSequenceClassification, AutoTokenizer
import torch
import torch.nn.functional as F

model_path = "./phobert_finetuned"
model = AutoModelForSequenceClassification.from_pretrained(model_path)
tokenizer = AutoTokenizer.from_pretrained(model_path)

model.eval()

def predict_label(text):
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding="max_length", max_length=128)

    with torch.no_grad():
        outputs = model(**inputs)

    logits = outputs.logits
    probs = F.softmax(logits, dim=-1)

    predicted_label = torch.argmax(probs, dim=-1).item()

    label_mapping = {0: "Trẻ em", 1: "Bình đẳng giới", 2: "Dân số", 3: "Hôn nhân và gia đình", 4: "Phòng, chống bạo lực gia đình", 5: "Khác"}

    return label_mapping[predicted_label], predicted_label+1, probs.tolist()
```

```python
text = "Những ai có thể yêu cầu trợ giúp pháp lý khi bị bạo lực gia đình?"
label, label_index, probabilities = predict_label(text)

print(f"Label dự đoán: {label}")
print(f"Index của label: {label_index}")
print(f"Xác suất từng nhãn: {probabilities}")
```
